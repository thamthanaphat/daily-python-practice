```python
# day 4
# data set

students = [
    {"name": "Anan", "score": 85},
    {"name": "Bua", "score": 92},
    {"name": "Chai", "score": 78},
    {"name": "Anan", "score": 85}  # duplicate entry
]


# 1.Removes the exact duplicate entry
unique_students = []
for i in students:
    if i not in unique_students:
        unique_students.append(i)

print(unique_students)

# 2.Creates a list of just the names
names = []
for i in students:
    names.append(i['name'])
print(names)

# 3.Creates a dictionary mapping name → score ({"Anan": 85, ...})
dic_name = {}
for i in students:
    dic_name[i['name']] = i['score']

print(dic_name)

# 4.Finds and prints the student with the highest score using the dictionary
best_score = max(dic_name, key = dic_name.get)
print(best_score,dic_name[best_score])

# 5.Prints the final unique count of students
len(unique_students)



# ----------------------------------------

inventory = [
    {"item": "Rice", "qty": 50, "price": 25},
    {"item": "Egg", "qty": 100, "price": 5},
    {"item": "Milk", "qty": 30, "price": 18},
]

# 1.Creates a list of just item names
item_names = []
for i in inventory:
    item_names.append(i['item'])
print(item_names)

# 2.Creates a dictionary mapping item → price
item_price = {}
for i in inventory:
    item_price[i['item']] = i['price']
print(item_price)

# 3.Finds the item with the lowest price using min() (like we did with max() for scores)
lower_price = min(item_price, key =item_price.get)
print(lower_price,item_price[lower_price])

# 4.Calculates total inventory value: sum(qty * price) for all items
total_inventory = 0
for i in inventory:
    total_inventory += i['qty'] * i['price']
    print(total_inventory)


# -------------------------------------
nums = [3, 5, 3, 8, 5, 9, 1, 8]

# Using set() — note: order may change
print(list(set(nums)))


# Using a for loop + if x not in new_list (like your dictionary duplicate fix, but simpler)
unique_nums = []
for i in nums:
    if i not in unique_nums:
        unique_nums.append(i)

print(unique_nums)

# Using a dictionary trick: list(dict.fromkeys(nums)) — this keeps order AND removes duplicates
print(list(dict.fromkeys(nums)))




#----------------------
sentence = "the quick brown fox jumps over the lazy dog the fox runs"

# Splits the sentence into a list of words (.split())
word_splits = sentence.split()
print(word_splits)

# Builds a dictionary counting how many times each word appears
word_counts = {}
for i in sentence.split():
    word_counts[i] = word_counts.get(i , 0) + 1
print(word_counts)


# Finds the most frequent word using max() with key=
max(word_splits,key = word_splits.count)

# Prints all words that appear more than once
unique_words = []
dup_words = []
for i in sentence.split():
    if i not in unique_words:
        unique_words.append(i)
    else:
        if i not in dup_words:
            dup_words.append(i)
print(dup_words)
```

```python
class BankAccount():
  def __init__(self,owner,balance=0):
      self.owner = owner
      self.balance = balance
  def deposit(self,amount):
      self.balance += amount
  def withdraw(self,amount):
    if self.balance <  amount:
        print('มันจะถอนเงินเยอะกว่าที่มีได้ไง')
    else:
        self.balance -= amount


  def show_balance(self):
      print(f"{self.owner}'s balance: {self.balance}")

tul = BankAccount('tul',200)
mild = BankAccount('mild',400)

mild.withdraw(600)
tul.deposit(200)
mild.show_balance()
tul.show_balance()
```

```python
class Vehicle:
    def __init__(self, plate, driver):
        self.plate = plate
        self.driver = driver
        self.status = "พร้อมใช้งาน"

    def report_breakdown(self, reason):
        self.status = "เสีย"
        print(f"🚨 {self.plate} ({self.driver}) เสีย: {reason}")

    def describe(self):
        return f"รถ {self.plate} คนขับ {self.driver} สถานะ {self.status}"
```
